3. Loops

```python
palabras = cadena1.strip(', ').split()
palabras
```

```python
for palabra in palabras:
    print(palabra)
```

```python
#Colocar numeracion a cada palabra
```

```python
for idx, palabra in enumerate(palabras):
    print(idx, palabra)
```
